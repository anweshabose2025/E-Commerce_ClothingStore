# Problem solved by pydantic

```python
from pydantic import BaseModel
class Student(BaseModel):
    name:str
    score:int
student1 = Student(name = "Anwesha", score = 33)
print(student1)
```

# Pydantic AI

```python
import nest_asyncio
nest_asyncio.apply() # we did this because pydantic_ai is an asynchronous framework. 
# asynchronous: things not taking place together.

import os
from dotenv import load_dotenv
from pydantic_ai import Agent
load_dotenv()
```

```python
basic_agent = Agent("groq:llama-3.1-8b-instant")
response = basic_agent.run_sync("Why is the sky blue, give reply in a funny way , in 10 words")
print("agent resposne is :" , response.output)
```

```python
basic_agent = Agent("groq:llama-3.1-8b-instant")
response = await basic_agent.run("Why is the sky blue, give reply in a funny way , in 10 words")
print("agent resposne is :" , response.output)
```

# ONE AGENT WITH PERSONNA
